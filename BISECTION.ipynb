```python
import math

# Fungsi persamaan
def f(x):
    return x**3 - x - 2

# Batas interval
a = 1
b = 2

# Toleransi error
tol = 0.0001

# Proses metode Bisection
if f(a) * f(b) >= 0:
    print("Interval tidak memenuhi syarat.")
else:
    iterasi = 1

    print("Iterasi\t a\t\t b\t\t c\t\t f(c)")

    while abs(b - a) > tol:
        c = (a + b) / 2

        print(iterasi, "\t", round(a, 4), "\t",
              round(b, 4), "\t", round(c, 4), "\t",
              round(f(c), 6))

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

        iterasi += 1

    print("\nAkar persamaan =", round(c, 6))
```

Program ini digunakan untuk mencari akar persamaan menggunakan metode Bisection (Bagi Dua) dengan menentukan interval awal, menghitung titik tengah, lalu mempersempit interval secara berulang hingga mencapai batas toleransi yang ditentukan.
