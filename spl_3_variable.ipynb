```python
# PENYELESAIAN SISTEM PERSAMAAN LINEAR 3 VARIABEL
# Menggunakan Metode Eliminasi Gauss
# Mata Kuliah: Komputasi Numerik

print("PROGRAM PENYELESAIAN SPL 3 VARIABEL")
print("METODE ELIMINASI GAUSS")
print("=" * 45)
```

```python
# Persamaan:
# 2x + y + z = 10
# x + 2y + z = 11
# x + y + 2z = 12

import numpy as np

A = np.array([
    [2, 1, 1],
    [1, 2, 1],
    [1, 1, 2]
], dtype=float)

B = np.array([10, 11, 12], dtype=float)

print("Matriks koefisien:")
print(A)

print("\nMatriks konstanta:")
print(B)
```

```python
# Proses eliminasi Gauss

n = len(B)

for k in range(n - 1):

    # Memilih baris pivot terbesar
    pivot = k + np.argmax(np.abs(A[k:, k]))

    # Menukar baris jika diperlukan
    A[[k, pivot]] = A[[pivot, k]]
    B[[k, pivot]] = B[[pivot, k]]

    # Mengubah elemen di bawah pivot menjadi nol
    for i in range(k + 1, n):
        faktor = A[i, k] / A[k, k]

        A[i, k:] = A[i, k:] - faktor * A[k, k:]
        B[i] = B[i] - faktor * B[k]

print("Matriks setelah eliminasi Gauss:")
print(A)
print("\nHasil konstanta:")
print(B)
```

```python
# Menghitung nilai variabel dengan substitusi mundur

X = np.zeros(n)

for i in range(n - 1, -1, -1):
    X[i] = (B[i] - np.dot(A[i, i+1:], X[i+1:])) / A[i, i]

print("HASIL PENYELESAIAN:")
print("Nilai x =", X[0])
print("Nilai y =", X[1])
print("Nilai z =", X[2])
```

```python
# Memeriksa apakah hasil memenuhi persamaan awal

A_awal = np.array([
    [2, 1, 1],
    [1, 2, 1],
    [1, 1, 2]
], dtype=float)

B_awal = np.array([10, 11, 12], dtype=float)

hasil = np.dot(A_awal, X)

print("Pemeriksaan hasil:")
print("Hasil perkalian matriks:", hasil)
print("Konstanta awal:", B_awal)

if np.allclose(hasil, B_awal):
    print("Kesimpulan: Solusi benar!")
else:
    print("Kesimpulan: Solusi belum sesuai!")
```
